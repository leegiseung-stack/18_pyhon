```python
#3의배수일때 피즈(Fizz) 5(Buzz)의 배수일때 버즈
for i in range(1,20+1):
    if i % 3 == 0:
        print("Fizz")
    elif i%5 ==0:
        print("Buzz")
    else :
        print(i)
```

```python
for i in range(1,20+1):
    if i%3 ==0 and i%5 ==0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i%5 ==0:
        print("Buzz")
    else :
        print(i)
```

```python
#코드단축하기
for i in range(1,30+1):
    print("Fizz" * (i%3==0) + "Buzz" * (i%5==0) or i)
```

```python
first , second = map(int,input().split())
for i in range(first,second+1):
    if i %35 == 0:
        print("FizzBuzz",end="")
    elif i % 5 == 0:
        print("Fizz",end="")
    elif i % 7 == 0:
        print("Buzz")
    else:
        print(i,end="")
```
