```python
# #for와 range사용하기

# for 변수 in range(횟수):
#     반복할 코드

for i in range (5):
    print("hello world")
```

```python
#반복문에서 변수의 변화 확인
for i in range (10):
    print("hello world",i) 
```

```python
for number in range (10):
    print("hello world",number) 
```

```python
#시작하는 숫자와 끝나는 숫자 지정하기 
for a in range(5,12):
    print(a)
```

```python
#증가폭을 이용해서 사용하기
for b in range(0,10,2):
    print(b)
```

```python
#숫자를 감소시키기 
for c in range(10,0,-1):
    print(c)
```

```python
for i in reversed(range(10)):
    print(i)
```

```python
#입력하는 횟수대로 반복하기 
count = int(input())
for i in range(count):
    print("hello world!",i)
```

```python
#시퀀스 객체로 바꾸기 
a=[10,20,30,40,50]
for i in a :
    print(i)
```

```python
#문자열
fruits=["apple","orange","grape"]
for fruit in fruits:
    print(fruit)
```

```python
for letter in "Python":
    print(letter, end=' ')
```

```python
for letter in reversed ("python"):
    print(letter,end="")
```

```python
#p193 리스트의 요소에 10을 곱해서 출력하기
x=[49,-17,25,102,8,62,21]
for i in x:
    print(i*10, end=' ')
```

```python
qqd=int(input())
print(qqd)

for i in range(1,10):
    print(qqd,i,sep="*",end="=")
    print(qqd*i)
    
```

```python
dan = int(input())
for num in range(1,10):
    print(f"{dan}*{num}={dan*num}")
```
