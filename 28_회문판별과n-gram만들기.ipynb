```python
#회문판별하기
#반복문으로 문장 검사하기
word = input("단어를 입력하세요")

is_palindrome=True #무조건 회문판단초기값
for i in range(len(word)//2) :         #단어갯수의 반만큼만돌아감
    if word[i]!=word[-1 -i]: 
        is_palindrome= False
        break

print(is_palindrome)
```

```python
#시퀀스 뒤집기로 문자 검사하기
word = input()
print(word == word[::-1])
```

```python
#n-gram만들기
##반복문으로 n-gram출력하기

text = "Hello" 
for i in range(len(text)-1):
    print(text[i],text[i+1],sep="")
```

```python
text="this is python script"
words = text.split()
words
for i in range(len(words)-1):
    print(words[i],words[i+1])
```

```python
#zip 2-gram 만들기
text = "hello"
two_gram = zip(text,text[1:])
list(two_gram)

for i in two_gram:
    print(i[0],i[1],sep="")
```

```python
#zip과 리스트 표현식으로 N-gram 만들기
text = "hello"
[text[i:]for i in range(5)]
```
