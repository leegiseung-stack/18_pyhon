```python
#유닛23번 심사문제
matrix = []
col, row = map(int, input().split())

for i in range(row):
    matrix.append(list(input()))

result = [[0 for 가로 in range(col)] for 세로 in range(row)] 
#미리 2차원배열울 만들어두기

for r in range(row):
    for c in range(col):
        if matrix[r][c] == "*": #지뢰검사
            result[r][c] = "*"  #지뢰면 그대로 결과값에 출력 
        else:
            count = 0
            # 위쪽
            if r - 1 >= 0 and matrix[r - 1][c] == "*":
                count += 1
            # 아래쪽
            if r + 1 < row and matrix[r + 1][c] == "*":
                count += 1
            # 왼쪽
            if c - 1 >= 0 and matrix[r][c - 1] == "*":
                count += 1
            # 오른쪽
            if c + 1 < col and matrix[r][c + 1] == "*":
                count += 1
            # 왼쪽 위
            if r - 1 >= 0 and c - 1 >= 0 and matrix[r - 1][c - 1] == "*":
                count += 1
            # 오른쪽 위
            if r - 1 >= 0 and c + 1 < col and matrix[r - 1][c + 1] == "*":
                count += 1
            # 왼쪽 아래
            if r + 1 < row and c - 1 >= 0 and matrix[r + 1][c - 1] == "*":
                count += 1
            # 오른쪽 아래
            if r + 1 < row and c + 1 < col and matrix[r + 1][c + 1] == "*":
                count += 1

            result[r][c] = str(count)

for r in result:
    print(*r, sep="")
    #print(r)     # ['*', '2', '*', '2', '1']
```

```python
col,row = map(int,input().split())
for i in range(row):
     matrix.append(list(input()))

# 2차원 리스트 초기화 오류 수정
result = [[0 for _ in range(col)] for _ in range(row)]

# 8방향 탐색을 위한 오프셋 (행 변화량, 열 변화량)
dr = [-1, -1, -1, 0, 0, 1, 1, 1]
dc = [-1, 0, 1, -1, 1, -1, 0, 1]

for r in range(row):
  for c in range(col):
    if matrix[r][c] == '*':
      result[r][c] = '*'
    else:
      count = 0
      # 8방향을 돌며 지뢰('*') 개수 세기
      for i in range(8):
        nr = r + dr[i]
        nc = c + dc[i]

        # 범위를 벗어나지 않고, 해당 위치가 지뢰('*')인 경우
        if 0 <= nr < row and 0 <= nc < col:
          if matrix[nr][nc] == '*':
            count += 1

      result[r][c] = count

# 결과 출력
for row_data in result:
  print(''.join(map(str, row_data)))
```

```python
#유닛30 연습문제
#가장 높은 점수를 구하는 함수 만들기
korean,english,mathematics,science=100,86,81,91

def get_max_score(*args):
    return max(args)
    
max_score=get_max_score(korean,english,mathematics,science)
print("높은점수:",max_score)

max_score=get_max_score(english,science)
print("높은점수:",max_score)
```

```python
#유닛30 심사문제 
korean,english,mathematics,science =map (int,input().split())

def get_min_max_score(*args):
    return min(args),max(args)
    
def get_average(*args):
    average_score=sum(args)/len(args)
    return average_score


min_score,max_score = get_min_max_score(korean,english,mathematics,science )
average_score = get_average(korean,english,mathematics,science)
print("낮은점수:{0:.2f}, 높은점수:{1:.2f}, 평균점수{2:.2f}".format(min_score,max_score,average_score))


min_score,max_score = get_min_max_score(english,science )
average_score = get_average(english,science)
print("낮은점수:{0:.2f}, 높은점수:{1:.2f}, 평균점수{2:.2f}".format(min_score,max_score,average_score))
```
