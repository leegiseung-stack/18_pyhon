```python
#유닛19 심사문제 별출력하기 
a=int(input())
for i in range(a):
    # 공백 출력 (줄어듦)
    for _ in range(a-i):
        print(" ", end="")
    # 별 출력 (홀수 개씩 늘어남)
    for k in range(2 * i + 1):
        print("*", end="")
    print()
```

```python
#유닛23번 심사문제
matrix = []
col, row = map(int, input().split())

for i in range(row):
    matrix.append(list(input()))

result = [[0 for 가로 in range(col)] for 세로 in range(row)] 
#미리 2차원배열울 만들어두기

for r in range(row):
    for c in range(col):
        if matrix[r][c] == "*": #지뢰검사
            result[r][c] = "*"  #지뢰면 그대로 결과값에 출력 
        else:
            count = 0
            # 위쪽
            if r - 1 >= 0 and matrix[r - 1][c] == "*":
                count += 1
            # 아래쪽
            if r + 1 < row and matrix[r + 1][c] == "*":
                count += 1
            # 왼쪽
            if c - 1 >= 0 and matrix[r][c - 1] == "*":
                count += 1
            # 오른쪽
            if c + 1 < col and matrix[r][c + 1] == "*":
                count += 1
            # 왼쪽 위
            if r - 1 >= 0 and c - 1 >= 0 and matrix[r - 1][c - 1] == "*":
                count += 1
            # 오른쪽 위
            if r - 1 >= 0 and c + 1 < col and matrix[r - 1][c + 1] == "*":
                count += 1
            # 왼쪽 아래
            if r + 1 < row and c - 1 >= 0 and matrix[r + 1][c - 1] == "*":
                count += 1
            # 오른쪽 아래
            if r + 1 < row and c + 1 < col and matrix[r + 1][c + 1] == "*":
                count += 1

            result[r][c] = str(count)

for r in result:
    print(*r, sep="")
    #print(r)     # ['*', '2', '*', '2', '1']
```
