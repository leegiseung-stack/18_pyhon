```python
#문자열 여러 줄을 파일에 쓰기,읽기
#반복문으로 파일여러번쓰기
with open("./data/Hello.txt","w") as file:
    for i in range(3):
        file.write(f"Hello,world{i} \n")
        
```

```python
#파일객체.writelines(문자열리스트)
lines = ["안녕하세요 \n","파이썬 \n","코딩도장입니다\n"]
with open("./data/Hello.txt","w") as file:
    file.writelines(lines)
```

```python
#한줄씩 리스트로 가져오기
with open("./data/Hello.txt","r") as file:
    lines = file.readline()
    print(lines)
```

```python
#파일의 내용을 한줄씩 읽기
with open("./data/Hello.txt","r") as file:
    line = None 
    while line != "":
        line = file.readline()
        print(line.strip("\n"))
```
