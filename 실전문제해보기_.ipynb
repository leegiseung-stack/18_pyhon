```python
#p94

# korean = 92
# english = 47
# mathematics = 86
# science = 81

korean,english,mathematics,science = map(int,input().split())
#한과목이라도 50점 미만이면 불합격
합격판정=(korean>50 and english>50 and mathematics>50 and science>50 )
if 합격판정==True :
    print("합격입니다")
else :
    print("불합격입니다")
```
