```python
#딕셔너리의 키의 자료형 
#키에는 리스트와 딕셔너리는 사용할수없음. -> 변경이가능해서
#키로는 정수,실수,불도 사용할수있음
x={
    100:"hundred",
    False:0,
    3.5:[3.5,3.5] # 벨류값으로는 리스트가능 키로는 불가능

}

print(x)
```
