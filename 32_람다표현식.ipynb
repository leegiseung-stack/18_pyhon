```python
#filter 
#filter(함수,반복가능한객체)

def f(x):
    return x >5 and x <10
a=[8,3,2,10,15,7,1,9,0,11]
#f(8)
#f(11)
list(filter(f,a))#함수명만적어줌 ()쓰지마
```

```python
from functools import reduce
#reduce(함수,반복가능한객체)
def f(x,y):
    return x+y

f(1,2)
a=[1,2,3,4,5]
reduce(f,a) #두개를 꺼내서 x,y에들어감 retrun은 한개
            #리턴된값 다음값을 더해서 retrun해줌
            #누적해서더해줌
```

```python
a=[1,2,3,4,5]
from functools import reduce
#import functools 도가능
reduce(lambda x,y:x+y,a)
```

```python
files=['font','1.png','10.jpg','11.gif','2.jpg','3.png','table.xslx','spec.docx']

a=lambda x: x.find('jpg') != -1 or x.find('png') != -1 
#find함수로 찾았을때  존재하면 인덱스로 숫자가반환 없으면 -1  
print(list(filter(a,files)))
```
