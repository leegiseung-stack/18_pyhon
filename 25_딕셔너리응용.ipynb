```python
#딕셔너리에서 키 값 수정하기 
x={
    "a":10,
    "b":20,
    "c":30,
    "d":40
}
x.update(a=90)
print(x)
x.update(e=50)
print(x)
```

```python
#반복문으로 딕셔너리의 키-값 쌍을 모두 출력하기
x={
    "a":10,
    "b":20,
    "c":30,
    "d":40
}
for i in x:
    print(i) #키값만가져옴

for key,value in x.items():
    print(key)
    print(value)
```

```python
#딕셔너리 표현식 사용하기
#{키:값 for 키,값 in 딕셔너리}
keys = ["a","b","c","d"]
imsi = dict.fromkeys(keys)
imsi

{key:value for key,value in imsi.items()} #키와 벨류사이에는 콜론이다
```

```python
#딕셔너리 표현식에서 if조건문 사용하기 
x= {"a":10,"b":20,"c":30,"d":40}
for key, value in x.items():
    if value ==20:
        del x[key]

#RuntimeError: dictionary changed size during iteration 
#딕셔너리 돌리는와중에 크기가 달라져서 에러남 
```

```python
x= {"a":10,"b":20,"c":30,"d":40}
y=x.copy()

for key, value in y.items():
    if value ==20:
        del x[key]
print(x)
print(y)
```
