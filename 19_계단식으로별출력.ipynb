```python
for i in range(5):
    # 공백 출력 (줄어듦)
    for j in range(4 - i):
        print(" ", end="")
    # 별 출력 (홀수 개씩 늘어남)
    for k in range(2 * i + 1):
        print("*", end="")
    print()
```

```python
for i in range(5):
    # 공백 출력 (늘어남)
    for j in range(i):
        print(" ", end="")
    # 별 출력 (줄어듦)
    for k in range(2 * (5 - i) - 1):
        print("*", end="")
    print()
```

```python
#사각형으로 별 출력하기 
for _ in range(5):
    for _ in range(5):
        print("*",end="")
    print()
```
