```python
#문자열 서식지정자와 포멧팅 사용하기
#서식 지정자로 문자열 넣기 
#%s % "문자열"
"I am %s" % "james" 
name="maria"
"i am %s" % name
```

```python
#서식 지정자로 숫자 넣기 
#"%d" % 숫자
"I am %d" %20

b = 20
"I am %d" %b
```

```python
#서식지정자로 소수점 표현하기
#"%f % 숫자(실수)"
"%f" % 2.3
```

```python
#"%.자리수 f" %숫자 
"%.2f"  %2.3

"%5.2f"  %2.3
```

```python
#서식지정자로 문자열 정렬하기
#%길이s 

"%10s" % "python"
```

```python
#문자열 왼쪽 정렬하기 
"%-10s" % "python"
```

```python
#서식 지정자로 문자열 안에 값 여러개 넣기
# "%d %s" % (숫자,문자)
"today is %d %s" % (3, "Aapril")
```

```python
#format 매서트 사용하기
print("hello, {0}".format("wrold"))
```

```python
#format메서드로 값을 여러개 넣기
print("hello {0} {2} {1}".format("python","script", 3.6))
```

```python
#format 메서드로 같은 값을 여러개 넣기
print("{0} {0} {1} {1}".format("python","script"))
```

```python
#format 매서드에서 인덱스 생략하기
print("Hello {} {} {}".format("python","script",3.6))
```

```python
#format매서드에서 인덱스대신 이름 지정해주기
print("Hello {language} {version}".format(language="python",version=3.6))
```
