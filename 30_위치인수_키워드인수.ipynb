```python
#위치인수를 사용하는 함수를 만들고 호출하기 
def print_numbers(a,b,c):
    print(a)
    print(b)
    print(c)

print_numbers(10,20,30)
```

```python
#언패킹 사용하기
x=[10,20,30]
print_numbers(x) #인수안에 [10,20,30]이 들어있는것
```

```python
#언패킹 사용하기
x=[10,20,30]
print_numbers(*x) #인수안에 [10,20,30]이 들어있는것
# *리스트를 풀어주는것!
```

```python
#p388
#가변인수 만들기 
#def 함수이름(*매개변수): *가변적
#코드

def print_numbers(*args): #변수는 아무거나 무관함
    for arg in args:
        print(arg)

print_numbers(10)
print("==="*10)
print_numbers(10,20,30,40)
```

```python
def print_numbers(*args): #가변인수
    print(args)
    print(type(args))

print_numbers(10,20,30)
```

```python
print_numbers([1,2,3,4,5])
#튜플안에 리스트의 객체 한번만나옴 
print_numbers(*[1,2,3,4,5])
#언팩킹
```

```python
#36 연습문제
#가장높은 점수를 구하는 함수 만들기
korean,english,mathematics,science=100,86,81,91

def get_max_score(*args):
    return max(args)
    
max_score=get_max_score(korean,english,mathematics,science)
print("높은점수:",max_score)

max_score=get_max_score(english,science)
print("높은점수:",max_score)
```

```python
korean,english,mathematics,science =map (int,input().split())

def get_min_max_score(*args):
    return min(args),max(args)
    
def get_average(*args):
    average_score=sum(args)/len(args)
    return average_score


min_score,max_score = get_min_max_score(korean,english,mathematics,science )
average_score = get_average(korean,english,mathematics,science)
print("낮은점수:{0:.2f}, 높은점수:{1:.2f}, 평균점수{2:.2f}".format(min_score,max_score,average_score))


min_score,max_score = get_min_max_score(english,science )
average_score = get_average(english,science)
print("낮은점수:{0:.2f}, 높은점수:{1:.2f}, 평균점수{2:.2f}".format(min_score,max_score,average_score))
```
