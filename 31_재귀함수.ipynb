```python
#재귀호출로 팩토리얼 구하기
def factrorial(n):
    if n==1:
        return 1 
    return n*factrorial(n-1)

print(factrorial(5))
#5*factorial(4)
#5*4*factrorial(3)
#5*4*3*factrorial(2)
#1: 5*4*3*2*1
```

```python
#31 연습문제
#재귀호출로 회문 판별하기 
def is_palindrome(word):
    if len(word)<2:
        return True
    if word[0] != word[-1]:
        return False
    return is_palindrome(word[1:-1])

print(is_palindrome("hello"))
print(is_palindrome("level"))
```

```python
#심사문제31
#재귀호출로파보나치 수 구하기 
a=[]
def fib(n):
  # 기저 조건: 0번째는 0, 1번째는 1
  if n == 0:
    a[n]=0
    return 0
  elif n == 1:
    a[1]=1
    return 1
    

  # 점화식 적용 및 반환
  a[n]=n
  return fib(n - 1) + fib(n - 2)


#n = int(input())
print(fib(5))
print(a)
```

```python
# n+1 크기만큼 0으로 채워진 리스트 미리 생성 (IndexError 방지)
a = [0] * (n + 1)


def fib(n):
  # 기저 조건
  if n == 0:
    a[0] = 0
    return 0
  elif n == 1:
    a[1] = 1
    return 1

  # 이미 계산된 값이 리스트에 있다면 그 값을 바로 재사용
  if a[n] != 0:
    return a[n]

  # 점화식으로 계산한 결과를 리스트에 저장하고 반환
  a[n] = fib(n - 1) + fib(n - 2)
  return a[n]

n = int(input())

print(fib(n))  # 5번째 피보나치 수 출력 (5)
print(a)  # 수열이 담긴 리스트 출력 ([0, 1, 1, 2, 3, 5])
```
