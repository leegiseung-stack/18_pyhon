```python
import random

# Yes와 No를 담은 리스트
options = ["Yes", "No"]

# 리스트에서 무작위로 하나를 선택
result = random.choice(options)
print(result)

# result = random.choice(options)

# # 결과 출력
# print(result)
```

```python
import random
# 레인지 객체 만들기
ran_ = range(10)
result = random.choice(ran_)
# ran_ 대신 result 변수를 사용하여 홀/짝 판별
if result % 2 == 0:
  print("O")  # 짝수일 때
else:
  print("X")  # 홀수일 때 (elif 대신 else를 쓰면 더 깔끔합니다)
#홀 
```
