```python
#문자열 판단하기
#특정문자열이 포함되어있는지 판단하기
#내장라이브러리
import re
re.match("hello","hello,world!")
re.match("python","hello,world!")
re.match("python","hello,world!")
```

```python
#문자열이 맨 앞에 오는지 맨 뒤에 오는지 판단하기
#^문자열 
#문자열$ #마치는거 찾기
#re.search("패턴","문자열")
re.search("^Hello","Hello,world")
re.search("world!$","Hello,world!")
```

```python
#지정된 문자열이 하나라도 포함되는지 판단하기
re.match("hello|world","hello")
```

```python
#범위 판단하기
#[0-9]* #0~9 하나의 숫자 
#[0-9]+ #0~9까지 하나 이상의 숫자가 있어야함

print(re.match("[0-9]*","1234"))
print(re.match("[0-9]+","1234"))
print(re.match("[0-9]+","abcd"))
```

```python
print(re.match("a*b","b")) #*는 하나도 없어도됨 
print(re.match("a+b","b")) #+는 한개이상있어야함 
print(re.match("a*b","aab")) #*는 하나도 없어도됨 
print(re.match("a*b","aaaaaaaaa")) #*는 하나도 없어도됨 
print(re.match("a+b","aab")) #*는 하나도 없어도됨 

```

```python
#문자가 한개만 있는지 판단하기 
#문자?
#[0-9]? 문자가 0개인지 1개인지 판단하기

print(re.match("abc?d" , "abd"))
print(re.match("abc?d" , "abcccccd"))
print(re.match("ab[0-9]?c" , "ab3c")) #? 한개있어도되고 0개있어도됨
print(re.match("ab.d","abxd")) #한개는 무조건있어야함
```

```python
#문자의 갯수 판단하기
#문자{개수}
#{문자열}{개수}
print(re.match("h{3}","hhhelllo"))
print(re.match("h{3}","hhhhhelllo"))
print(re.match("hello{3}","hellohellohelloworld"))
print(re.match("(hello){3}","hellohellohelloworld")) #hello를 그룹으로 묶어줌
```

```python
re.match("[0-9]{3}","011-1234-1234") #반드시숫자가 세개가있어야함
re.match("[0-9]{3}-[0-9]{4}-[0-9]{4}", "010-1234-1234") #010-1234-1234
print(re.match("[0-9]{3}-[0-9]{4}-[0-9]{4}", "010-1234-124") )#010-1234-1234
```

```python
#(문자){시작개수,끝개수}
#(문자열){시작개수,끝개수}
#[0-9]{시작개수,끝개수}

print(re.match("[0-9]{2,3}-[0-9]{3,4}-[0-9]{4}","02-123-1234"))
print(re.match("[0-9]{2,3}-[0-9]{3,4}-[0-9]{4}","010-1234-1234"))
print(re.match("[0-9]{2,3}-[0-9]{3,4}-[0-9]{4}","02-12-1234"))
```

```python
#숫자와 영문 문자를 조합해서 판단하기
#a-z 
#A-Z
print(re.match("[a-zA-Z0-9]+" ,"hello1234"))# [ ]는 한개의 글자 영문 소 ,영문 대 ,0-9 가능 +(한개이상)
print(re.match("[A-Z0-9]+" ,"hello1234"))# [ ]는 한개의 글자 영문 대 ,0-9 가능 +(한개이상)
```

```python
#가-힣 (가장마지막단어 힣)
print(re.match("[가-힣]+","홍길동")) #한개이상
```

```python
#특정문자 범위에 포함되지 않는지 판단하기
#[^범위] # [] 대괄호 안에 있는^ 범위가아닌것 찾기
        # () 안에 있는 ^ 범위 찾기
print(re.match("[^A-Z]+" , "Hello"))
print(re.match("[^a-z]+" , "Hello"))
```

```python
import re

#^[범위] 시작할때
print(re.match("^[A-Z]+" ,"Hello0")) #대문자로 시작할때를 찾는거
#[범위]$ 끝날때 
print(re.match("[0-9]+$" ,"Hello0"))
```
