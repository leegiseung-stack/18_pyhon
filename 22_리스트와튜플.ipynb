```python
#리스트의 요소 하나 추가하기 
a = [10,20,30]
a.append(500)
print(a)

len(a)
```

```python
a=[]
a.append(10)
print(a)
```

```python
#리스트안에 리스트 추가하기
a=[10,20,30]
a.append([500,600])
print(a)
len(a)
```

```python
#리스트 확장하기
a=[10,20,30]
a.extend([500,600])
print(a)
len(a)
```

```python
#리스트에 특정인덱스에 요소 추가하기
a=[10,20,30]
a.insert(2,500)
print(a)
```

```python
a=[10,20,30]
a.insert(1,700)
print(a)
```

```python
a=[10,20,30]
a.insert(len(a),500) #항상 마지막을 가르킴
print(a)
```

```python
a=[10,20,30]
a.insert(1,[500,600])
print(a)
```

```python
a=[10,20,30]
a[1:1] = [500,600]
print(a)
```

```python
a=[10,20,30]
a[1:2] = [500,600]
print(a)

#위와 다르게 원래들어가있던 20사라짐
```

```python
#리스트에서 특정인덱스에 요소 삭제하기 
a=[10,20,30]
print(a.pop()) #꺼내진 마지막값을 출력
print(a)
```

```python
a=[10,20,30]
print(a.pop(1))
print(a)
```

```python
a=[10,20,30]
del a[1]
print(a)
```

```python
#리스트에서 특정값을 찾아서 삭제하기 
a=[10,20,30]
print(a.remove(20)) #해당하는 값으로 지움 #그냥지울뿐 팝처럼 값이 들어가진않음
print(a)
```

```python
a=[10,20,30,20]
a.remove(20) #해당하는 값으로 지움 #첫번째것만 지워짐
print(a)
```
