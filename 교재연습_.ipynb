```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score
        
    def check_pass(self):
        # 60점 이상이면 합격, 아니면 불합격 반환
        if self.score >= 60:
            return f"{self.name}님은 합격입니다."
        else:
            return f"{self.name}님은 불합격입니다."

# 사용 예시
student1 = Student("김철수", 75)
print(student1.check_pass())
# 출력 결과: 김철수님은 합격입니다.
```

```python
3. 리스트/튜플과 반복문: 조건에 맞는 요소 필터링
특정 조건(예: 짝수만 골라내기 등)을 만족하는 요소들만 새로운 리스트나 튜플에 담아 반환하는 로직입니다.

Python
def get_even_tuple(numbers):
    even_list = []
    for num in numbers:
        if num % 2 == 0:  # 짝수 조건
            even_list.append(num)
            
    # 리스트를 튜플로 변환하여 반환
    return tuple(even_list)

# 사용 예시
nums = (1, 2, 3, 4, 5, 6, 7, 8)
print(get_even_tuple(nums))
# 출력 결과: (2, 4, 6, 8)
```

```python
4. 딕셔너리 정렬: 값(Value) 또는 키(Key) 기준으로 정렬하기
딕셔너리를 점수나 빈도수(value) 기준으로 정렬하거나 이름(key) 기준으로 정렬하는 코드는 응용 문제로 매우 자주 나옵니다.

Python
def sort_dict_by_value(d):
    # 딕셔너리의 아이템을 값(x[1]) 기준으로 내림차순 정렬
    sorted_items = sorted(d.items(), key=lambda x: x[1], reverse=True)
    return sorted_items

# 사용 예시
scores = {'국어': 80, '수학': 95, '영어': 85}
print(sort_dict_by_value(scores))
# 출력 결과: [('수학', 95), ('영어', 85), ('국어', 80)]
```

```python
# 입력 예시: 철수 80 영희 95 민수 70 지영 85
inputs = input().split()
students = []

# 2칸씩 건너뛰며 (이름, 점수) 튜플을 만들어 리스트에 저장
for i in range(0, len(inputs), 2):
    name = inputs[i]
    score = int(inputs[i+1])
    students.append((name, score))

# 평균 점수 계산 (점수들의 총합 / 학생 수)
total_score = 0
for name, score in students:
    total_score += score
average = total_score / len(students)

# 평균 점수 이상인 학생의 이름만 추출
result = []
for name, score in students:
    if score >= average:
        result.append(name)
```

```python
# 변형 유형 1: 1차원 리스트 지뢰찾기 (양옆 2방향 검사)
# i-1과 i+1 인덱스가 리스트 범위를 넘지 않는지(0 이상, 전체 길이 미만) 확인하고 지뢰(*) 개수를 세는 코드입니다.

Python
def count_1d_mines(arr):
    result = []
    n = len(arr)
    
    for i in range(n):
        # 이미 지뢰인 칸은 그대로 유지
        if arr[i] == '*':
            result.append('*')
            continue
            
        mine_count = 0
        
        # 1. 왼쪽 칸 확인 (i - 1)
        if i - 1 >= 0 and arr[i - 1] == '*':
            mine_count += 1
            
        # 2. 오른쪽 칸 확인 (i + 1)
        if i + 1 < n and arr[i + 1] == '*':
            mine_count += 1
            
        result.append(str(mine_count))
        
    return result

# 사용 예시
print(count_1d_mines(['.', '*', '.', '*', '.']))
# 출력 결과: ['1', '*', '2', '*', '1']
```

```python
# 변형 유형 2: 안전지역(길)의 최대 연속 길이 구하기
# 지뢰를 만나기 전까지 연속된 길(O)이 최대 몇 칸인지 구하는 전형적인 카운팅 패턴입니다. 지뢰(X)를 만나면 카운트가 초기화됩니다.

Python
def max_safe_length(arr):
    max_count = 0
    current_count = 0
    
    for item in arr:
        if item == 'O':  # 길을 만나면
            current_count += 1
            # 지금까지의 최대값과 비교해서 더 크면 갱신
            if current_count > max_count:
                max_count = current_count
        else:  # 지뢰(X)를 만나면 연속 끊김
            current_count = 0
            
    return max_count

# 사용 예시
print(max_safe_length(['O', 'O', 'X', 'O', 'O', 'O', 'X', 'O']))
# 출력 결과: 3 (세 번째 칸의 'O, O, O'가 가장 김)
```

```python
def count_4dir_mines(grid):
    rows = len(grid)
    cols = len(grid[0])
    
    # 결과를 담을 똑같은 크기의 2차원 리스트 생성
    result = [[0] * cols for _ in range(rows)]
    
    for r in range(rows):
        for c in range(cols):
            if grid[r][c] == '*':
                result[r][c] = '*'
                continue
                
            count = 0
            
            # 1. 위쪽 칸 검사 (r - 1, c)
            if r - 1 >= 0 and grid[r - 1][c] == '*':
                count += 1
                
            # 2. 아래쪽 칸 검사 (r + 1, c)
            if r + 1 < rows and grid[r + 1][c] == '*':
                count += 1
                
            # 3. 왼쪽 칸 검사 (r, c - 1)
            if c - 1 >= 0 and grid[r][c - 1] == '*':
                count += 1
                
            # 4. 오른쪽 칸 검사 (r, c + 1)
            if c + 1 < cols and grid[r][c + 1] == '*':
                count += 1
                
            result[r][c] = count
            
    return result

# 사용 예시
grid = [
    ['.', '*', '.'],
    ['.', '.', '.'],
    ['*', '.', '.']
]
# 결과 확인용 출력
for row in count_4dir_mines(grid):
    print(row)
```
