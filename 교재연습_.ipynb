```python
def print_numbers(*args):
    for arg in args:
        print(arg)

print_numbers(10)
print("======="*3)
print_numbers(10,20,30,40)
print("======="*3)
print_numbers(*[10])
print("======="*3)
print_numbers(list(range(10)))
#단 이때 def print_number(*args ,a ) 처럼 *args가 고정매개변수보다 앞쪽에 오면 안됨! 
#매개변수 순서상 *args는 반드시 뒤쪽에 와야 합니다 


```

```python
#키워드 인수 사용하기
def personal_info(name,age,address):
    print("이름:",name),
    print("나이:",age),
    print("주소:",address)

personal_info("홍길동",30,"율도국")
print("=========="*5)
personal_info(name="홍홍홍",address="율율율",age=31)
#키워드 인수를 활용하면 순서를 맞추지 않아도 해당하는 값에 들어감! 
```

```python
def personal_info(name,age,address):
    print("이름:",name),
    print("나이:",age),
    print("주소:",address)

x={'name':"홍길동",'age':30,"address":"율도국"}
personal_info(**x)
print("=========="*5)
personal_info(**{'name':"홍길동",'age':30,"address":"율도국"})
print("=========="*5)
personal_info(*x)
print("=========="*5)
#personal_info(**{'name':"홍길동",'age':30,"address":"율도국","취미":"도술"})
#없는 변수값을 넣으면 오류가남
```

```python
def personal_info(**kwargs):
    for kw,arg in kwargs.items():
        print(kw,':',arg,sep="")
personal_info(name="홍길동")
print("=="*30)
personal_info(name="홍길동",age=30,address='서울시 용산구 이촌동')
```

```python
def personal_info(**kwargs):
    for kw,arg in kwargs.items():
        print(kw,':',arg,sep="")

y={
    'name':"홍길동",'age':30,"address":"율도국"
}


def personal_info_find(**kwargs):
    for kw, arg in kwargs.items():
        # 키(kw)가 'name'일 때만 실행되도록 조건을 추가합니다.
        if kw == 'name':
            print("{0}이 등록되어있습니다".format(arg))
       # if '홍길동' == kwargs["name"]:
       #     print("{0}이 등록되어있습니다".format[arg])
       # else : print("등록되어있지않습니다")
        if kw=='age':
            print("{0}세 입니다".format(arg))
personal_info(**y)
print("=="*30)
personal_info_find(**y)
```
