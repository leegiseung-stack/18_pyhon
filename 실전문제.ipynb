```python
#유닛14 p174

국어,영어,수학,과학=map(int,input().split())

if 0<=국어<=100 or 0<=영어<=100 or 0<=수학<=100 or 0<=과학<=100:
    avg=(국어+영어+수학+과학)/4
    if avg>80:
        print("합격")
    else : 
        print("불합격")
else :
    print("잘못된 점수")
```
