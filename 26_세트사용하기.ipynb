```python
# 집합연산사용하기
## 합집합
##세트1 | 세트2 #합집합
#set.union(세트1,세트2) => 세트1.union(세트2)

a={1,2,3,4}
b={3,4,5,6}

print(a|b)
print(set.union(a,b))
print(a.union(b))
```

```python
#교집합
## 세트1 & 세트2
## set.intersection(a,b)
## a(intersection(b))


a={1,2,3,4}
b={3,4,5,6}

print(a & b)
print(set.intersection(a,b))
print(a.intersection(b))
```

```python
#차집합
## a-b
## set.difference(a,b) 
## a.difference(b)


a={1,2,3,4}
b={3,4,5,6}
print(a-b)
print(set.difference(a,b))
print(a.difference(b))
print("="*20)
print(set.difference(b,a))
```

```python
#대칭차집합
##세트1 ^ 세트2
##set.symmetric_difference(a.b)

a={1,2,3,4}
b={3,4,5,6}
print(a^b)
print(set.symmetric_difference(a,b))
#둘다속하지않는것(교집합의반대)
```

```python
#집합연산후 할당연산자 사용하기 
##세트1 |= 세트2 (합집합후에 재할당)
##세트1.update(세트2)

a={1,2,3,4}
a |={5}
print(a)
a={1,2,3,4}
a.update({5})
print(a)
```

```python
#intersection_update
##세트1&= 세트2
###intersection_update
##세트1.intersection_update(세트2)

a={1,2,3,4}
a &= {0,1,2,3,4}
print(a)

a={1,2,3,4}
a.intersection_update({0,1,2,3,4})
print(a)
```

```python
#부분집합 상위집합인지 확인하는 개념
## 현재세트 <=다른세트
## 현재세트.is  (인지아닌지 T,F반환하는함수만들때)
## 현재세트.issubset(다른세트) (포함되어있는지아닌지)
a={1,2,3,4}
a<={1,2,3,4}
a.issubset({1,2,3,4})
```

```python
#세트표현식사용하기
##{식 for 변수 in 반복가능한객체}
##set{식 for 변수 in 반복가능한객체}
a={i for i in "apple"}
print(a)
a={i :0 for i in "apple"}
print(a)
```

```python
a= {i for i in "pineapple" if i not in "apl"}
print(a)

a= {i :0 for i in "pineapple" if i not in "apl"}
print(a)
```
