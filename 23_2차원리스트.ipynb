```python
###########################################
col, row = map(int,input().split())
matrix = []
for _ in range(row):
    matrix.append((list(input())))

for i in range(row):
    for j in range(col):
        matrix[i][j] == "."
        count = 0 
        #카운트갯수추가해야함   
        for r in range(i-1,i+2):
            for c in range(j-1,j+2):

                if matrix[i][j] == "*":
                    matrix[r][c] = "*"
                if 0<= r < 3 and 0<= c < col:
                    if matrix[r][c] == "*":
                        count+=1


        matrix[i][j] = count

# print(col,row)

print(*matrix,sep="\n")

```

```python
col, row = map(int, input().split())
matrix = []
for _ in range(row):
    matrix.append(list(input()))

# 결과를 저장할 새로운 격자 생성
result_matrix = [[0] * col for _ in range(row)]

for i in range(row):
    for j in range(col):
        # 1. 현재 칸이 지(*)뢰라면 그대로 '*' 유지
        if matrix[i][j] == "*":
            result_matrix[i][j] = "*"
        else:
            # 2. 지뢰가 아니라면 주변 3x3 범위의 '*' 개수 세기
            count = 0 
            for r in range(i - 1, i + 2):
                for c in range(j - 1, j + 2):
                    # 범위 체크 (0 <= r < row, 0 <= c < col)
                    if 0 <= r < row and 0 <= c < col:
                        if matrix[r][c] == "*":
                            count += 1
            result_matrix[i][j] = count

# 출력 형식 맞추기 (리스트의 각 행을 문자열로 합쳐서 출력)
for r in result_matrix:
    print("".join(map(str, r)))
```
