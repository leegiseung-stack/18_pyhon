```python
result =0
def add(num):
    global result
    result += num
    return result

print(add(5))
print(add(10))
print(add(2))
```

```python
# #속성사용하기 
# class 클래스이름:
#     def __init__(self): #변수선언
#         self.속성명=값#변수명

class Person:
    def __init__(self):
        self.hello ="안녕하세요"
    def greeting(self):
        print(self.hello)

james = Person()
print(james.hello) #속성호출할땐 필요x
james.greeting() #함수호출할땐 괄호
```

```python
#인스턴스를 만들때 값 받기 
# class 클래스이름:
#     def __init__(self,매개변수1,매개변수2):
#         self.속성1=매개변수1
#         self.속성2=매개변수2

class Person:
    def __init__(self,name,age,address):
        self.hello="안녕하세요"
        self.name= name #self.name은 내맘대로 변수는 변할수있음
        self.age=age
        self.address = address

    def greeting(self):
        print(f"{self.hello} 저는 {self.name}입니다")

maria=Person("마리아",28,"서울시 서초구 반포동")
maria.greeting()
        
```
