```python
#global 전역변수
x=10
def foo():
    global x #밖에서도 바뀜 #코드길어지면 헷갈려서 추천안함
    x=20
    print(x)

foo()
print(x)
```

```python
def foo():
    global a 
    a=20
    print(a)

foo()
print(a)
```

```python
def print_helllo():
    hello = "hello,world!"
    def print_message():
        print(hello)
    print_message()

print_helllo()
```

```python
#지역변수의 범위 
#지역변수 변경하기
def A():
    x=10
    def B():
        x=20

    B()
    print(x)

A()
```

```python
#nonlocal 지역변수
def A():
    x=10
    def B():
        nonlocal x #가장가까운x를 전역변수로쓰겠다.
        x=20

    B()
    print(x)

A()
```

```python
#nonlocal이 변수를 찾는 순서
def A():
    x=10
    y=100
    def B():
        x=20
        def C():
            nonlocal x
            nonlocal y
            x=x+30
            y=y+300
            print(x)
            print(y)
        C()
    B()
    
    print(x)
    print(y)
A()
```

```python
#global로 전역변수 사용하기 
x=1
def A():
    x=10
    def B():
        x=20
        def C():
            global x
            x=x+30
            print(x)
        C()
    B()
A()
```

```python
#클로져 사용하기 :
def calc():
    a= 3
    b= 5
    def mul_add(x):
        return a*x+b
        
    return mul_add #함수명만 가지고와서 아래쪽에 괄호넣고 인수넣어주면 실행됨
c=calc()
print(c(1),c(2),c(4),c(5))
```

```python
#lambda로 클로져 만들기
def calc():
    a=3
    b=5
    return lambda x:a*x +b

c= calc()
print(c(1))
```

```python
#시험x
##클로져의 지역변수 변경하기
def calc():
    a=3
    b=5
    total=0
    def mul_add(x):
        nonlocal total 
        total =total+a*x +b
        print(total)
    return mul_add

c=calc()
c(1)
c(2)
```
