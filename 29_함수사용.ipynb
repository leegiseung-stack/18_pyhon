```python
#return으로 함수 중간에서 빠져나오기
def not_ten(a):
    if a == 10:
        return #return None 생략
    print(a,"입니다",sep="")

not_ten(5)
not_ten(10)
```

```python
x,y=map(int,input().split())

def calc(num_1,num_2):
    add=float(num_1+num_2)
    sub=float(num_1-num_2)
    mul=float(num_1*num_2)
    div=float(num_1/num_2)
    return add,sub,mul,div
a,s,m,d=calc(x,y)

print("덧셈{0},뺄셈{1},곱셈{2},나눗셈{3}".format(a,s,m,d))
```
